````python
need_for_JC_correction = """
Raw Hamming distance only counts the positions that currently differ
between two sequences. It can miss mutations that happened at the same
position more than once. For example, a position could change from A to
C and then from C to G. It would look like only one mutation occurred, 
but two actually did. The Jukes-Cantor correction estimates the true 
number of mutations from the observed differences. It accounts for 
these hidden or reversed mutations. For closely related SARS-CoV-2 
genomes, the correction is small because multiple mutations at the same 
position are rare.
"""
````

````python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from Bio import SeqIO

from helper_functions import jukes_cantor


def estimate_mutation_rate(fasta_path, exclude_reference=True):
    records = list(SeqIO.parse(fasta_path, "fasta"))

    reference_record = records[0]
    reference_seq = str(reference_record.seq).upper()

    parsed = []

    for record in records:
        fields = record.description.split("|")

        record_id = fields[0]
        country = fields[1]
        date = fields[2]

        parsed.append({
            "id": record_id,
            "description": record.description,
            "country": country,
            "date": date,
            "sequence": str(record.seq).upper(),
            "is_reference": record.id == reference_record.id
        })

    df = pd.DataFrame(parsed)

    df["collection_date"] = pd.to_datetime(
        df["date"],
        format="mixed"
    )

    # Time elapsed since the earliest collected sequence.
    start_date = df["collection_date"].min()
    df["days_since_start"] = (
        df["collection_date"] - start_date
    ).dt.days

    # JC-corrected distance from the first genome/reference.
    df["jc_distance"] = df["sequence"].apply(
        lambda seq: jukes_cantor(reference_seq, seq)
    )

    # The reference has zero distance and zero elapsed time.
    df.loc[df["is_reference"], "jc_distance"] = 0.0
    df.loc[df["is_reference"], "days_since_start"] = 0

    # Exclude the reference from the regression fit.
    if exclude_reference:
        fit_df = df[~df["is_reference"]]
    else:
        fit_df = df

    slope, intercept = np.polyfit(
        fit_df["days_since_start"],
        fit_df["jc_distance"],
        deg=1
    )

    return df, slope, intercept, len(reference_seq)


# Ebola virus
ebola_df, ebola_slope, ebola_intercept, ebola_genome_length = (
    estimate_mutation_rate("data/p2-ebola-viruses.fasta")
)

ebola_per_genome_per_day = ebola_slope
ebola_per_nt_per_day = ebola_slope / ebola_genome_length

print("Ebola virus")
print(f"Mutation rate per genome per day: {ebola_per_genome_per_day:.2g}")
print(f"Mutation rate per nucleotide per day: {ebola_per_nt_per_day:.2g}")

# Zika virus
zika_df, zika_slope, zika_intercept, zika_genome_length = (
    estimate_mutation_rate("data/p2-zika-viruses.fasta")
)

zika_per_genome_per_day = zika_slope
zika_per_nt_per_day = zika_slope / zika_genome_length

print("\nZika virus")
print(f"Mutation rate per genome per day: {zika_per_genome_per_day:.3g}")
print(f"Mutation rate per nucleotide per day: {zika_per_nt_per_day:.3g}")
````

````python
def plot_mutation_rate(df, slope, intercept, title, filename):
    fig, ax = plt.subplots(figsize=(8, 5))

    # Plot non-reference genomes.
    non_reference = df[~df["is_reference"]]
    reference = df[df["is_reference"]]

    ax.scatter(
        non_reference["days_since_start"],
        non_reference["jc_distance"],
        s=25,
        alpha=0.75,
        label="Sampled genomes"
    )

    # Show the reference at the origin, but do not include it in the fit.
    ax.scatter(
        reference["days_since_start"],
        reference["jc_distance"],
        color="black",
        s=45,
        label="Reference genome"
    )

    x_fit = np.linspace(0, df["days_since_start"].max(), 200)
    y_fit = slope * x_fit + intercept

    ax.plot(
        x_fit,
        y_fit,
        color="red",
        linewidth=2,
        label="Linear regression"
    )

    ax.set_xlabel("Days since earliest sampled genome")
    ax.set_ylabel("Jukes-Cantor corrected genetic distance")
    ax.set_title(title)
    ax.legend()
    ax.grid(alpha=0.25)

    fig.tight_layout()
    fig.savefig(filename)
    plt.show()


plot_mutation_rate(
    ebola_df,
    ebola_slope,
    ebola_intercept,
    "Zaire ebolavirus genetic divergence over time",
    "problem2_ebola.svg"
)

plot_mutation_rate(
    zika_df,
    zika_slope,
    zika_intercept,
    "Zika virus genetic divergence over time",
    "problem2_zika.svg"
)
````

````python
reference_in_regression = """
Why did we ignore the reference when fitting the regression line? Keep 
in mind that in our biological interpretation, this is not an outlier 
in the machine learning sense.

We ignored the reference because it is the earliest known sequence and 
is placed at the origin by definition. Including it would assume that all 
genetic distance starts from zero at that date. However, Ebola and Zika were 
already circulating before the reference was sampled, so other sequences can 
have a large distance from it even early in the dataset. This creates a positive
intercept rather than a line through the origin. The reference is not an 
outlier. It is simply not the common ancestor of all sampled viruses, so excluding 
it prevents the regression from being incorrectly forced through zero.
"""
````

````python
comparing_mutation_rates = """
We should compare mutation rates on a per nucleotide basis. A per genome rate 
depends on genome length: a longer genome can accumulate more mutations per day
simply because it has more nucleotides. A per nucleotide rate divides by genome 
length, so it measures the average mutation rate of each nucleotide. This makes 
comparisons between viruses with different genome sizes fair. Per genome rates 
describe total change in one virus, but per nucleotide rates are better for 
comparing different viruses.
"""
````

## Problem 3: Reinterpreting mutation rates

Majority of RNA viral genomes are single-stranded RNA, where only one strand of the viruses encode for the genes. We divide these viruses into positive-sense and negative-sense viruses, based on where they code for genes.
Positive-sense RNA directly codes for genes and acts like mRNA, meaning it can be directly used by the host's ribosomes to make proteins.
Negative-sense RNA is a reverse complement of the positive-sense RNA and does not code for genes directly. It must first be converted into positive-sense RNA by the virus's own enzymes before it can be used to make proteins.
Positive-sense RNA viruses mutate faster because their replication process is simpler and mostly lacks error correction. Negative-sense RNA viruses, on the other hand, mutate more slowly since they have an extra step in replication and require more functional genes to support replication.

## Problem 4: Variant-specific mutations

As the virus mutates, it inevitably evolves and proliferates around the world. Every so often, some mutations may prove especially beneficial to the spread of the virus, and this version of the virus spreads faster than other versions. When a version of a virus becomes especially prevalent inside a population, we call this a virus *variant*. Variants are nothing more than a naming scheme for viruses that have specific mutations. For instance, in Slovenia during 2022, we were dealing with the Omicron variant. Think of this as observing natural selection in real time. Some viruses have mutations that enable them to spread more easily throughout our population, which inevitably leads to the demise of other virus variants, which are not as good at proliferation. The result is survival of the fittest at the viral level, where, unfortunately, the fittest viruses seem to cause the most damage to us humans.

How do we identify variants? A variant is determined by several so-called *defining mutations*. Mutations can either be synonymous or nonsynonymous. Synonymous mutations are changes in nucleotide bases that result in the same encoded amino acid and are thus less important. Nonsynonymous mutations are nucleotide mutations that alter the amino acid sequence of a protein.

## Problem 5: Identifying variants

The second part is comprised of the actual mutation. For instance, *S:T 19 R* means that we are looking at a mutation on the S -- spike protein, where the original amino acid at location 19 was T -- threonine, which changed to R -- arginine. The "-" symbol indicates a deletion (e.g., *S:H 69 -*). For a complete reference of SARS-CoV-2 mutations, you can take a look at https://covariants.org and look through the different variants.
````python
import json
import re
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from Bio import SeqIO

# Load the reference spike protein and Australian spike sequences
reference_fasta = "data/NC_045512.2.fasta"
spike_fasta = "data/p5-aus-spike-proteins.fasta"
variants_json = "variants.json"

reference_record = next(SeqIO.parse(reference_fasta, "fasta"))
reference_seq = str(reference_record.seq).upper()

records = list(SeqIO.parse(spike_fasta, "fasta"))

# Load variant definitions
variants_json = "data/variants.json"
with open(variants_json) as f:
    variants_data = json.load(f)

# Parse mutations 
mutation_pattern = re.compile(r"^(?:S:)?([A-Z\-])(\d+)([A-Z\-])$")

def parse_mutation(mutation_string):
    match = mutation_pattern.match(mutation_string.strip())

    if match is None:
        raise ValueError(f"Could not parse mutation: {mutation_string}")

    reference_aa = match.group(1)
    position = int(match.group(2))
    variant_aa = match.group(3)

    return reference_aa, position, variant_aa


# Extract all substitutions and deletions from one spike sequence
def get_sequence_mutations(sequence):
    mutations = set()

    for index, (reference_aa, sequence_aa) in enumerate(
        zip(reference_seq, sequence),
        start=1
    ):
        # Ignore ambiguous sequencing positions.
        if sequence_aa == "X":
            continue

        # Ignore insertions; pre-aligned sequences are assumed to have none.
        if sequence_aa == "-":
            continue

        if sequence_aa != reference_aa:
            if sequence_aa == "-":
                mutation = f"{reference_aa}{index}-"
            else:
                mutation = f"{reference_aa}{index}{sequence_aa}"

            mutations.add(mutation)

    return mutations


# Classify each sequence
classification = {}

# for record in records[:5]:
#     print(record.description)

for record in records:
    fields = record.description.split("|")
    sequence_id = fields[0]
    collection_date = pd.to_datetime(fields[1], format="mixed")

    sequence = str(record.seq).upper()
    sequence_mutations = get_sequence_mutations(sequence)

    # Sequence identical to the reference spike protein.
    if len(sequence_mutations) == 0:
        classification[sequence_id] = {
            "date": collection_date,
            "variant": "NO_MUTATIONS"
        }
        continue

    best_variant = "UNKNOWN"
    best_match_percentage = 0.0

    for variant_name, variant_info in variants_data.items():
        variant_mutations = variant_info.get("nonsynonymous", [])

        if not variant_mutations:
            continue

        matching_mutations = 0

        for mutation_info in variant_mutations:
            if isinstance(mutation_info, str):
                mutation_string = mutation_info
            else:
                mutation_string = (
                    f"{mutation_info['left']}"
                    f"{mutation_info['pos']}"
                    f"{mutation_info['right']}"
                )

            _, position, variant_aa = parse_mutation(mutation_string)

            # Convert 1-based amino-acid position to 0-based Python index.
            index = position - 1

            # Ignore ambiguous positions.
            if index >= len(sequence) or sequence[index] == "X":
                continue

            if sequence[index] == variant_aa:
                matching_mutations += 1

        match_percentage = matching_mutations / len(variant_mutations)

        if match_percentage > best_match_percentage:
            best_match_percentage = match_percentage
            best_variant = variant_name

    # Require at least 51% matching mutations.
    if best_match_percentage >= 0.51:
        assigned_variant = best_variant
    else:
        assigned_variant = "UNKNOWN"

    classification[sequence_id] = {
        "date": collection_date,
        "variant": assigned_variant
    }

# Save classification CSV
classification_df = pd.DataFrame(
    [
        {"accession_id": sequence_id, "variant": details["variant"]}
        for sequence_id, details in classification.items()
    ]
)

classification_df.to_csv(
    "problem5-classification.csv",
    index=False,
    header=False
)

# Check the required CSV format
check_df = pd.read_csv("problem5-classification.csv", sep=",", header=None)
print(check_df.head())
print(check_df.value_counts())
````

````python
## Monthly distribution plot
# Prepare monthly variant percentages
monthly_records = []

for sequence_id, details in classification.items():
    monthly_records.append({
        "date": details["date"],
        "variant": details["variant"]
    })

monthly_df = pd.DataFrame(monthly_records)

# Count sequences by month and variant
monthly_counts = (
    monthly_df
    .groupby([
        pd.Grouper(key="date", freq="MS"),
        "variant"
    ])
    .size()
    .unstack(fill_value=0)
)

# Convert counts to percentages within each month
monthly_percentages = monthly_counts.div(
    monthly_counts.sum(axis=1),
    axis=0
) * 100

# Ensure UNKNOWN and NO_MUTATIONS are included if present
for variant_name in ["UNKNOWN", "NO_MUTATIONS"]:
    if variant_name in monthly_percentages.columns:
        continue

# Sort variants by their overall maximum percentage for a cleaner plot
variant_order = (
    monthly_percentages
    .max()
    .sort_values(ascending=False)
    .index.tolist()
)

monthly_percentages = monthly_percentages[variant_order]

# Stacked area plot
fig, ax = plt.subplots(figsize=(12, 6))

x_values = monthly_percentages.index
y_values = [monthly_percentages[variant].values for variant in variant_order]

ax.stackplot(
    x_values,
    y_values,
    labels=variant_order,
    alpha=0.85
)

ax.set_xlabel("Collection date")
ax.set_ylabel("Percentage of sequences (%)")
ax.set_title("SARS-CoV-2 variant distribution in Australia over time")
ax.set_ylim(0, 100)
ax.legend(
    loc="upper left",
    bbox_to_anchor=(1.01, 1),
    title="Variant"
)
ax.grid(alpha=0.25)

fig.tight_layout()
fig.savefig("problem5.svg")
plt.show()
````
